```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("StudentsPerformance.csv")
```

```python
df.head()
```

```python
df.info()
```

```python
df.describe()
```

Average Math Score

```python
avg_math = df["math score"].mean()

print(avg_math)
```

Highest Maths Scoring Students


```python
top_students = df.sort_values(
    by="math score",
    ascending=False
)

top_students.head()
```

```python
df["math score"].hist()

plt.xlabel("Math Score")
plt.ylabel("Students")
plt.title("Math Score Distribution")

plt.show()
```

Gender wise Performance

```python
df.groupby("gender")["math score"].mean()
```

Pass/Fail Column

```python
df["result"] = df["math score"].apply(lambda x: "Pass" if x >= 40 else "Fail")
df["result"].value_counts()
```

```python
df["math score"].hist(bins=10)
plt.show()
```
